```python
# Import libraries
import cv2
import os
from google.colab.patches import cv2_imshow

# Video path
video_path = "/content/sample_data/30 Sec of infinity war _ iron man suit up.mp4"

# Open the video
vid = cv2.VideoCapture(video_path)

# Create folder to save frames
if not os.path.exists("data"):
    os.makedirs("data")

# Initialize frame counter
current_frame = 0

# Read and save each frame
while True:
    success, frame = vid.read()

    # Stop if no more frames
    if not success:
        break

    # Save the frame as an image
    cv2.imwrite(f"data/frame{current_frame}.jpg", frame)

    # Increase frame count
    current_frame += 1

# Release the video after extraction
vid.release()

# Print extraction details
print("Frame extraction completed!")
print("Total frames extracted:", current_frame)

# ----------------------------------------
# Display a specific frame (Example: Frame 100)
# ----------------------------------------

frame_number = 1

# Open the video again
vid = cv2.VideoCapture(video_path)

# Move to the desired frame
vid.set(cv2.CAP_PROP_POS_FRAMES, frame_number)

# Read that frame
success, frame = vid.read()

if success:
    print(f"\nDisplaying Frame {frame_number}")
    cv2_imshow(frame)
else:
    print(f"Frame {frame_number} not found.")

# Release the video
vid.release()
```
